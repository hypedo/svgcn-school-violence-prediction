```python
import torch
import torch.nn.functional as F
import numpy as np

from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, roc_auc_score

from torch_geometric.data import Data
from torch_geometric.nn import GCNConv
```

```python
# 학교 하나를 벡터하나로 보고, 그 벡터간의 거리 유사도를 통해서 가까운 15개 학교를 연결시킴, 총 15개의 이웃을 보는것임

def make_knn_edges_from_scaled_features(
    X_scaled,
    k=15 # 학교마다 가까운 15개 학교를 연결하겠다라는 뜻
):
    nbrs = NearestNeighbors( # KNN모델 생성
        n_neighbors=k + 1,
        metric="cosine" # 거리기준은 cosine으로 설정 -> 코사인 유사도는 벡터의 방향이 얼마나 비슷한지 보여줌 -> 전체적인 피쳐 패턴이 비슷하면 가까운 학교라고 생각
    )

    nbrs.fit(X_scaled) # knn학습

    distances, indices = nbrs.kneighbors(X_scaled)

    edge_set = set()

    for i in range(X_scaled.shape[0]):
        for j in indices[i][1:]:
            edge_set.add((i, j)) # 양방햔 엣지 -> 만약에 0에서 1을 연결시켰으면 1도 0을 연결시키게 만드는 코드 edge_set.add((i, j))
            edge_set.add((j, i)) #

    edge_index = torch.tensor(list(edge_set), dtype=torch.long).t().contiguous() # pytorch Geometric형식으로 변환해서 edge형식으로 바꿔줌

    return edge_index
```

```python
# 학교폭력_빈도수를 타겟컬럼으로 설정, -> 그렇기에 따로 빼고 17개의 피쳐로 학습할 수 있게끔 세팅

target_col = "학교폭력_빈도수"

feature_cols = [col for col in df.columns if col != target_col]

X = df[feature_cols].apply(pd.to_numeric, errors="coerce")

print("사용 피처 개수:", len(feature_cols))
print(feature_cols)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(df[feature_cols]) # 피쳐들 스케일링 시작

x = torch.tensor(X_scaled, dtype=torch.float) # 입력값들을 pytorch tensor로 바꿈

# 회귀 문제라 y는 float
y_raw = df[target_col].values.astype(float)
y_log = np.log1p(df[target_col].values.astype(float)) # 학교폭력 발생 빈도수 같은 데이터는 보통 값의 분포가 한쪽으로 치우칠 수 있다.
# -> 그러기에 log변환을 통해 큰 값을 압축

y = torch.tensor( # y값도 tensor로 바꿈
    y_log,
    dtype=torch.float
).view(-1, 1)

print(x.shape)
print(y.shape)
```

```python
# edge만드는 코드 (위에서 만든 함수 여기서 사용)

edge_index = make_knn_edges_from_scaled_features(
    X_scaled=X_scaled,
    k=15 # k값은 15로 설정
)

print("스케일링 피처 기반 edge_index shape:", edge_index.shape) # 총 3062개의 edge가 만들어진 것을 볼 수 있음
```

```python
# 지금까지 만든 입력 피쳐, edge, 타겟값을 하나로 묶어서 pytorch Geometric의 객체로 만드는 코드

data = Data(
    x=x,
    edge_index=edge_index,
    y=y
)

print(data) # 학교가 160개, 피쳐수 17개, edge수 3062개
```

```python
# 데이터를 학습용, 검증용, 테스트용으로 나누는 코드

num_nodes = data.num_nodes
indices = np.arange(num_nodes)

train_idx, temp_idx = train_test_split(
    indices,
    test_size=0.3, # 나머지 70프로 데이터 학습
    random_state=42
)

val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.5, # 30프로의 절반을 검증용, 테스트용으로 나눈다
    random_state=42
)

data.train_mask = torch.zeros(num_nodes, dtype=torch.bool)
data.val_mask = torch.zeros(num_nodes, dtype=torch.bool)
data.test_mask = torch.zeros(num_nodes, dtype=torch.bool)

data.train_mask[train_idx] = True
data.val_mask[val_idx] = True
data.test_mask[test_idx] = True

print("Train:", data.train_mask.sum().item())
print("Val:", data.val_mask.sum().item())
print("Test:", data.test_mask.sum().item())

# 그래서 밑에가 그 데이터 개수
```

```python
# GCN회귀 모델 코드

class GCNRegressor(torch.nn.Module):
    def __init__(self, input_dim, hidden_dim):
        super(GCNRegressor, self).__init__()

        self.conv1 = GCNConv(input_dim, hidden_dim)
        self.conv2 = GCNConv(hidden_dim, hidden_dim)
        self.conv3 = GCNConv(hidden_dim, 1) # 총 3개의 conv를 사용

    def forward(self, data):
        x, edge_index = data.x, data.edge_index

        x = self.conv1(x, edge_index)
        x = F.relu(x) # 활성화 함수로는 ReLU함수 사용
        x = F.dropout(x, p=0.2, training=self.training) # drop_out 비율은 0.2로 설정 -> 과적합을 지우기 위한 랜덤하게 20프로 뉴런들 꺼버리기

        x = self.conv2(x, edge_index)
        x = F.relu(x)
        x = F.dropout(x, p=0.2, training=self.training)

        x = self.conv3(x, edge_index)

        return x
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("사용 장치:", device)

data = data.to(device) # 데이터를 cuda로 옮기기

model = GCNRegressor( # 모델 설정
    input_dim=data.x.shape[1],
    hidden_dim=8 # 은닉층은 8로 설정
).to(device)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.0005, # learning rate는 0.005로 설정
    weight_decay=5e-4 # L2정규화라고 볼 수 있음
)

loss_fn = torch.nn.L1Loss() # loss함수는 L1loss를 사용 -> 예측값과 실제값의 절대 오차 평균을 계산한다.
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
def evaluate_regression(mask):
    model.eval()

    with torch.no_grad():
        out = model(data)

        # log 예측값을 원래 빈도수로 복원
        pred = torch.expm1(out)
        pred = torch.clamp(pred, min=0)

        y_true_log = data.y[mask].cpu().numpy().reshape(-1)
        y_pred = pred[mask].cpu().numpy().reshape(-1)

        y_true = np.expm1(y_true_log)

        mae = mean_absolute_error(y_true, y_pred) # 여기 밑에 애들은 평가지표
        rmse = np.sqrt(mean_squared_error(y_true, y_pred))
        r2 = r2_score(y_true, y_pred)

    return mae, rmse, r2
```

```python
def to_risk_level_array(y):
    y = np.array(y).reshape(-1)

    levels = []

    for v in y: # 범주 설정
        if v <= 1:
            levels.append(0) # 0~1이면 저위험
        elif v <= 4:
            levels.append(1) # 2~4이면 중위험
        else:
            levels.append(2) # 5이상이면 고위험군으로 분류

    return np.array(levels)


def evaluate_risk_classification(mask):
    model.eval()

    with torch.no_grad():
        out = model(data)

        pred = torch.expm1(out)
        pred = torch.clamp(pred, min=0)

        y_true_log = data.y[mask].cpu().numpy().reshape(-1)
        y_pred = pred[mask].cpu().numpy().reshape(-1)

        y_true = np.expm1(y_true_log)

    true_level = to_risk_level_array(y_true)
    pred_level = to_risk_level_array(y_pred)

    acc = accuracy_score(true_level, pred_level)
    cm = confusion_matrix(true_level, pred_level)

    report = classification_report(
        true_level,
        pred_level,
        target_names=["저위험", "중위험", "고위험"],
        zero_division=0
    )

    return acc, cm, report, y_true, y_pred, true_level, pred_level
```

```python
for epoch in range(1, 1001): # 여기가 모델 학습 코드 epoch수는 1000번을 돌림
    model.train()
    optimizer.zero_grad()

    out = model(data)

    loss = loss_fn(
        out[data.train_mask],
        data.y[data.train_mask]
    )

    loss.backward()
    optimizer.step()

    if epoch % 50 == 0: # 50epoch마다 성능을 확인할 수 있게끔 해주는 코드
        train_mae, train_rmse, train_r2 = evaluate_regression(data.train_mask)
        val_mae, val_rmse, val_r2 = evaluate_regression(data.val_mask)

        print(
            f"Epoch {epoch:03d} | "
            f"Loss: {loss.item():.4f} | "
            f"Train MAE: {train_mae:.4f} | "
            f"Val MAE: {val_mae:.4f} | "
            f"Val RMSE: {val_rmse:.4f} | "
            f"Val R2: {val_r2:.4f}"
        )
```

```python
test_mae, test_rmse, test_r2 = evaluate_regression(data.test_mask)

print("Test MAE:", test_mae)
print("Test RMSE:", test_rmse)
print("Test R2:", test_r2)

risk_acc, risk_cm, risk_report, y_true, y_pred, true_level, pred_level = evaluate_risk_classification(data.test_mask)

print("\n위험도 Accuracy:", risk_acc)

print("\nConfusion Matrix:")
print(risk_cm)

print("\nClassification Report:")
print(risk_report)

# 여기가 평가 코드
# 여기 해석이 중요 -> 먼저 밑에 보면 베이스라인 코드가 있는데 MAE값이 2.88인것을 볼 수 있는데 우리 모델은 1.9까지 내렸음
# 테스트 데이터 24개 기준에서 위험도 분류 정확도 75%를 달성, 특히 고위험군에서 탐지 성능이 잘나옴
# 고위험인데 저위험군이라고 예측한 건수가 0개임
# 결정계수를 보면 매우 높은 설명력을 가지고 있다고 볼 수는 어렵지만 그래도 예측력이 있음
# 혼동행렬 해석
# 왼쪽이 실제 저,중,고이고 오른쪽 컬럼 하나하나가 예측값임 -> 그니까 3행의 2열보면 실제는 고위험군이지만 우리 모델은 중위험이라고 판단한게 3건이 있다는 뜻
# 학교폭력 예측할때는 고위험 학교를 저위험으로 놓치는 게 가장 위험한 오류임. 그런데 이 결과에서는 실제 고위험 학교를 저위험으로 예측한 경우가 없다.
# -> 이게 그나마 우리 모델에서 고무적인 부분임! 강조해야된다....
```
